```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import skimage
import tifffile
```

```python
image = tifffile.imread("/Users/fdrgsp/Desktop/t/frame_EColi_1_MMStack_Pos0_ph.tif")
labeled_mask = tifffile.imread(
    "/Users/fdrgsp/Desktop/t/EColi_2_MMStack_Pos0_labels.tif"
)
```

```python
fig, axis = plt.subplots(1, 2, figsize=(8, 4))
axis[0].imshow(image, cmap="gray")
axis[0].set_title("Original")
axis[0].axis("off")
axis[1].imshow(labeled_mask, cmap="nipy_spectral")
axis[1].set_title("Labeled Mask")
axis[1].axis("off")

plt.tight_layout()
plt.show()
```

```python
props_dict = skimage.measure.regionprops_table(
    labeled_mask,
    intensity_image=image,
    properties=[
        "label",
        "intensity_mean",
        "area",
        "eccentricity",
        "axis_major_length",
        "axis_minor_length",
    ],
)
print(props_dict.keys())
```

```python
props_df = pd.DataFrame(props_dict)
print(props_df)
```

```python
props_df["area"].plot(
    kind="hist",
    bins=200,
    title="Area Distribution",
    xlabel="Area (pixels)",
    ylabel="Frequency",
)
```

```python
props_df["eccentricity"].plot(kind="hist", bins=200, title="Eccentricity Distribution")
```

```python
props_df["axis_major_length"].plot(
    kind="hist", bins=200, title="Axis Major Length Distribution"
)
```

```python
props_df["axis_minor_length"].plot(
    kind="hist", bins=200, title="Axis Minor Length Distribution"
)
```
